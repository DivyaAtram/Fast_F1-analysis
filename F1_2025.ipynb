```python
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
import numpy as np
import fastf1 
```

```python
a = fastf1.get_event_schedule(2025)
a = a[['Country','Location','EventDate','EventName']]
```

```python
a
```

```python
r = fastf1.get_session(2025,'Dutch Grand Prix','R')
r.load()
r = r.results
r
```

```python
e= a.EventName.unique()
```

```python
final = pd.DataFrame(columns= ['TeamName','FullName','Position','Time','Status','Points','Laps','Venue'])
for i in e:
    s= fastf1.get_session(2025,i,'R')
    s.load()
    s=s.results
    s= s[['TeamName','FullName','Position','Time','Status','Points','Laps']]
    s['Venue'] = i
    final = pd.concat([final,s])
```

```python
final
```

```python
print(final.Venue.unique())
v= input("Enter Venue name:")
p= final[final.Venue == v]
plt.figure(figsize = (10,5))
sb.barplot(x=p.FullName, y=p.Points)
plt.xticks(rotation=90)
plt.show()
```

```python
final.to_csv("F1_2025_GP.csv")
```
